## Part A — Cleaning and Reshaping Data

### A messy dataset
Real data is rarely clean. Below is a small synthetic dataset with the kinds of problems
you'll see constantly: missing values, duplicate rows, and inconsistent text formatting.

```python
data = {
    'student':  ['Ana', 'ben ', 'Cara', 'Ana', 'David', 'ELI', None],
    'score':    [88, 72, None, 88, 65, 91, 74],
    'hours':    [5, 3, 4, 5, 2, 6, 3],
}
df = pd.DataFrame(data)
print(df)
```

### Cleaning steps
1. Drop exact duplicate rows.
2. Standardize text (strip whitespace, fix casing).
3. Handle missing values — here we drop rows missing a `student` name, but fill missing
   `score` with the column mean (a common, defensible default for numeric gaps).

```python
clean = df.drop_duplicates()
clean = clean.dropna(subset=['student'])
clean['student'] = clean['student'].str.strip().str.title()
clean['score'] = clean['score'].fillna(clean['score'].mean())

print(clean)
```

### Reshaping — wide vs. long
`melt` and `pivot` convert between "wide" (one column per variable) and "long"
(one row per observation) formats. Long format is usually what plotting libraries want.

```python
wide = pd.DataFrame({
    'student': ['Ana', 'Ben', 'Cara'],
    'quiz1':   [80, 70, 90],
    'quiz2':   [85, 75, 95],
})
long = wide.melt(id_vars='student', var_name='quiz', value_name='score')
print('Wide:'); print(wide)
print('\nLong:'); print(long)
```

## Part B — Visualizing Distributions and Relationships

```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
idf = iris.frame
idf['species'] = idf['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

fig, ax = plt.subplots(figsize=(7, 4))
for species, color in zip(['setosa', 'versicolor', 'virginica'], ['steelblue', '#c9a84c', '#ff6b6b']):
    subset = idf[idf['species'] == species]
    ax.hist(subset['petal length (cm)'], bins=12, alpha=0.6, label=species, color=color)
ax.set_title('Petal Length Distribution by Species')
ax.set_xlabel('Petal length (cm)'); ax.set_ylabel('Count')
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
colors = idf['target'].map({0: 'steelblue', 1: '#c9a84c', 2: '#ff6b6b'})
ax.scatter(idf['petal length (cm)'], idf['petal width (cm)'], c=colors, alpha=0.7)
ax.set_title('Petal Length vs. Width')
ax.set_xlabel('Petal length (cm)'); ax.set_ylabel('Petal width (cm)')
plt.show()
print('Notice: the three species roughly cluster — this is why petal size alone can classify Iris well.')
```
