## Part 1 — Gradient Descent Visualizer

```python
def run_gd(start=3.0, lr=0.2, n_steps=20):
    w = start
    history = [w]
    for _ in range(n_steps):
        w = w - lr * 2 * w   # gradient of w² is 2w
        history.append(w)
    return history

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
w_curve = np.linspace(-3.5, 3.5, 300)

for i, (lr, color) in enumerate([(0.05, 'steelblue'), (0.3, '#c9a84c'), (0.95, '#ff6b6b')]):
    h = run_gd(3.0, lr, 30)
    ax = axes[i]
    ax.plot(w_curve, w_curve**2, 'gray', linewidth=1.5)
    ax.plot([w for w in h], [w**2 for w in h], 'o-', color=color, markersize=5)
    ax.set_title(f'lr={lr}')
    ax.set_xlabel('w'); ax.set_ylabel('Loss = w²')

plt.suptitle('Gradient Descent: Effect of Learning Rate', fontsize=12)
plt.tight_layout()
plt.show()
print('lr=0.05: slow but stable  |  lr=0.3: efficient  |  lr=0.95: overshoots')
```

## Part 2 — Decision Trees (HOML Ch. 6)

```python
iris = load_iris()
X_tr, X_te, y_tr, y_te = train_test_split(iris.data, iris.target,
                                           test_size=0.2, random_state=42)

for depth in [2, 4, None]:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=42)
    tree.fit(X_tr, y_tr)
    train_acc = tree.score(X_tr, y_tr)
    test_acc  = tree.score(X_te, y_te)
    label = f'depth={depth}' if depth else 'depth=unconstrained'
    print(f'{label:30s}: train={train_acc:.3f}, test={test_acc:.3f}')

# Visualize a shallow tree
fig, ax = plt.subplots(figsize=(14, 5))
tree2 = DecisionTreeClassifier(max_depth=2, random_state=42)
tree2.fit(X_tr, y_tr)
plot_tree(tree2, feature_names=iris.feature_names,
           class_names=iris.target_names, filled=True, ax=ax)
ax.set_title('Decision Tree — depth=2 (Iris dataset)')
plt.show()
```

## Part 3 — First Peek at Neural Networks (HOML Ch. 10 intro)
A tiny 2-layer neural network on a non-linearly separable dataset (make_moons).
This is a preview — you'll go deeper after finishing the HOML arc.

```python
from sklearn.neural_network import MLPClassifier

X_m, y_m = make_moons(n_samples=500, noise=0.2, random_state=42)
X_m_tr, X_m_te, y_m_tr, y_m_te = train_test_split(X_m, y_m, test_size=0.2, random_state=42)

sc = StandardScaler()
X_m_tr = sc.fit_transform(X_m_tr)
X_m_te = sc.transform(X_m_te)

# Compare: linear (logistic) vs neural net
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression(); log_reg.fit(X_m_tr, y_m_tr)
mlp = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=500, random_state=42)
mlp.fit(X_m_tr, y_m_tr)

print(f'Logistic Regression test accuracy: {log_reg.score(X_m_te, y_m_te):.3f}')
print(f'Neural Network test accuracy:      {mlp.score(X_m_te, y_m_te):.3f}')
print()
print('The moons dataset is NOT linearly separable.')
print('Logistic regression fails; the neural network learns the curved boundary.')
```
