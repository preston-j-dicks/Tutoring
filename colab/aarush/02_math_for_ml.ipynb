## Week 3 — Functions, Vectors, Derivatives

### Plotting Functions
Before we minimize a function, we need to see it. Here's how to plot f(x) = x² and its derivative.

```python
x = np.linspace(-3, 3, 300)

f = x**2           # f(x) = x²
df = 2*x           # f'(x) = 2x (derivative)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(x, f, color='steelblue', linewidth=2.5, label='f(x) = x²')
axes[0].axhline(0, color='gray', linewidth=0.8)
axes[0].axvline(0, color='gray', linewidth=0.8)
axes[0].set_title('f(x) = x² (parabola — loss surface shape)')
axes[0].set_xlabel('x'); axes[0].set_ylabel('f(x)')
axes[0].legend()

axes[1].plot(x, df, color='#c9a84c', linewidth=2.5, label="f'(x) = 2x (derivative)")
axes[1].axhline(0, color='gray', linewidth=0.8)
axes[1].axvline(0, color='gray', linewidth=0.8)
axes[1].set_title("f'(x) = 2x — slope of the parabola")
axes[1].set_xlabel('x'); axes[1].set_ylabel("f'(x)")
axes[1].legend()

plt.tight_layout()
plt.show()

print('At x=2: f(2) =', 2**2, ', f\'(2) =', 2*2)
print('Positive derivative → function is rising. Go left to find minimum.')
```

## Week 4 — Gradients and Gradient Descent

```python
# 1D Gradient Descent — minimize f(w) = w²
# Gradient: df/dw = 2w
# Update: w ← w - α * gradient

def gradient_descent_1d(start=3.0, lr=0.2, n_steps=20):
    w = start
    history = [(w, w**2)]
    for _ in range(n_steps):
        grad = 2 * w     # derivative of w²
        w = w - lr * grad
        history.append((w, w**2))
    return history

history = gradient_descent_1d(start=3.0, lr=0.2, n_steps=20)

w_vals = np.linspace(-3.5, 3.5, 300)
loss_vals = w_vals**2

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(w_vals, loss_vals, 'steelblue', linewidth=2, label='Loss = w²')
axes[0].plot([h[0] for h in history], [h[1] for h in history],
              'o-', color='#c9a84c', markersize=6, linewidth=1.5, label='GD steps')
axes[0].set_title('Gradient Descent on w²'); axes[0].set_xlabel('w'); axes[0].set_ylabel('Loss')
axes[0].legend()

axes[1].plot([h[1] for h in history], 'o-', color='#ff6b6b', linewidth=2)
axes[1].set_title('Loss vs Iteration'); axes[1].set_xlabel('Step'); axes[1].set_ylabel('Loss')

plt.tight_layout()
plt.show()

print(f'Started at w=3.0, loss=9.0')
print(f'After 20 steps: w={history[-1][0]:.6f}, loss={history[-1][1]:.8f}')
```

### Gradient Descent for Linear Regression
Now with a real ML problem: find the best line through data points.

```python
np.random.seed(42)
X = 2 * np.random.rand(50, 1)
y = 3 * X + 5 + np.random.randn(50, 1)  # true: slope=3, intercept=5
X_b = np.c_[np.ones((50, 1)), X]         # add bias column

w = np.random.randn(2, 1)   # random init: [bias, slope]
lr = 0.1
n_epochs = 500
losses = []

for epoch in range(n_epochs):
    preds = X_b @ w
    residuals = preds - y
    mse = (residuals**2).mean()
    grad = (2 / len(X_b)) * X_b.T @ residuals  # MSE gradient
    w -= lr * grad
    losses.append(mse)

print(f'Learned intercept: {w[0,0]:.3f}  (true: 5.0)')
print(f'Learned slope:     {w[1,0]:.3f}  (true: 3.0)')

plt.figure(figsize=(8, 3))
plt.plot(losses, color='#c9a84c')
plt.title('Training Loss — Linear Regression via Gradient Descent')
plt.xlabel('Epoch'); plt.ylabel('MSE')
plt.show()
```
